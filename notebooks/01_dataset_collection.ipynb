### This notebook is for the loading of all datasets used in this project along with an initial analysis

```python
from datasets import load_dataset
import pandas as pd
```

```python
def inspect_dataset(dataset, name):
    print("=" * 80)
    print(f"DATASET: {name}")
    print("=" * 80)

    print("\nAvailable Splits:")
    print(dataset)

    for split in dataset.keys():
        print(f"\n{'-'*60}")
        print(f"Split: {split}")
        print(f"Number of samples: {len(dataset[split])}")

        print("\nFeatures:")
        print(dataset[split].features)

        print("\nFirst Example:")
        print(dataset[split][0])

    print("\n")
```

## 1. MRPC

```python
mrpc = load_dataset("glue", "mrpc")

inspect_dataset(mrpc, "MRPC")
```

## 2. QQP

```python
qqp = load_dataset("glue", "qqp")

inspect_dataset(qqp, "QQP")
```

## 3. PAWS

```python
paws = load_dataset("paws", "labeled_final")

inspect_dataset(paws, "PAWS")
```

## 4. MSCOCO Captions

```python
mscoco = load_dataset("sentence-transformers/coco-captions")

inspect_dataset(mscoco, "MSCOCO Captions")
```

## Overview

```python
datasets = {
    "MRPC": mrpc,
    "QQP": qqp,
    "PAWS": paws,
    "MSCOCO": mscoco
}

for name, ds in datasets.items():
    print("=" * 50)
    print(name)

    for split in ds.keys():
        print(f"{split}: {len(ds[split])} samples")
```
